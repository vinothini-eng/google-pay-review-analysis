```python
# ============================================================
# GOOGLE PAY INDIA DATA ANALYSIS
# STEP 2: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. LOAD CLEANED DATASET
# ------------------------------------------------------------
from google.colab import files

uploaded = files.upload()
df = pd.read_csv("GooglePayIndia_Cleaned.csv")

# Convert date columns
df['at'] = pd.to_datetime(df['at'], errors='coerce')
df['repliedAt'] = pd.to_datetime(df['repliedAt'], errors='coerce')

print("=" * 70)
print("GOOGLE PAY INDIA - EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nDataset Shape:")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nFirst 5 Rows:")
print(df.head())


# ------------------------------------------------------------
# 2. DATASET INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("1. DATASET INFORMATION")
print("=" * 70)

print(df.info())


# ------------------------------------------------------------
# 3. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("2. DESCRIPTIVE STATISTICS")
print("=" * 70)

print(df.describe())


# ------------------------------------------------------------
# 4. RATING DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("3. RATING DISTRIBUTION")
print("=" * 70)

rating_counts = df['score'].value_counts().sort_index()

print(rating_counts)

print("\nRating Percentage:")
rating_percentage = (
    df['score']
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print(rating_percentage)


# Rating Bar Chart
plt.figure(figsize=(8, 5))

rating_counts.plot(kind='bar')

plt.title("Google Pay Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. AVERAGE, MEDIAN AND MODE RATING
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("4. RATING STATISTICS")
print("=" * 70)

average_rating = df['score'].mean()
median_rating = df['score'].median()
mode_rating = df['score'].mode()[0]

print("Average Rating :", round(average_rating, 2))
print("Median Rating  :", median_rating)
print("Mode Rating    :", mode_rating)


# ------------------------------------------------------------
# 6. RATING-BASED SENTIMENT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("5. SENTIMENT ANALYSIS - RATING BASED")
print("=" * 70)


def classify_rating(score):
    if score <= 2:
        return "Negative"
    elif score == 3:
        return "Neutral"
    else:
        return "Positive"


df['sentiment'] = df['score'].apply(classify_rating)

sentiment_counts = df['sentiment'].value_counts()

print("Sentiment Counts:")
print(sentiment_counts)

sentiment_percentage = (
    df['sentiment']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\nSentiment Percentage:")
print(sentiment_percentage)


# Sentiment Chart
plt.figure(figsize=(8, 5))

sentiment_counts.plot(kind='bar')

plt.title("Google Pay Review Sentiment")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 7. CREATE YEAR AND MONTH COLUMNS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("6. DATE ANALYSIS")
print("=" * 70)

df['year'] = df['at'].dt.year
df['month'] = df['at'].dt.month
df['year_month'] = df['at'].dt.to_period('M')

print("Date range:")
print("From:", df['at'].min())
print("To  :", df['at'].max())


# ------------------------------------------------------------
# 8. YEARLY REVIEW COUNT
# ------------------------------------------------------------

yearly_reviews = df.groupby('year').size()

print("\nYearly Review Count:")
print(yearly_reviews)

plt.figure(figsize=(8, 5))

yearly_reviews.plot(kind='bar')

plt.title("Google Pay Reviews by Year")
plt.xlabel("Year")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. MONTHLY REVIEW COUNT
# ------------------------------------------------------------

monthly_reviews = df.groupby('year_month').size()

print("\nMonthly Review Count:")
print(monthly_reviews)


plt.figure(figsize=(12, 5))

monthly_reviews.plot(kind='line', marker='o')

plt.title("Google Pay Reviews Over Time")
plt.xlabel("Month")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 10. AVERAGE RATING BY MONTH
# ------------------------------------------------------------

monthly_rating = df.groupby('year_month')['score'].mean()

print("\nAverage Rating by Month:")
print(monthly_rating)


plt.figure(figsize=(12, 5))

monthly_rating.plot(kind='line', marker='o')

plt.title("Average Google Pay Rating by Month")
plt.xlabel("Month")
plt.ylabel("Average Rating")
plt.ylim(1, 5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 11. AVERAGE RATING BY YEAR
# ------------------------------------------------------------

yearly_rating = df.groupby('year')['score'].mean()

print("\nAverage Rating by Year:")
print(yearly_rating)


plt.figure(figsize=(8, 5))

yearly_rating.plot(kind='bar')

plt.title("Average Google Pay Rating by Year")
plt.xlabel("Year")
plt.ylabel("Average Rating")
plt.ylim(1, 5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 12. THUMBS-UP ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("7. THUMBS-UP ANALYSIS")
print("=" * 70)

print(df['thumbsUpCount'].describe())


# Top 10 most liked reviews
top_reviews = (
    df.sort_values(
        'thumbsUpCount',
        ascending=False
    )
    [['score', 'thumbsUpCount', 'content']]
    .head(10)
)

print("\nTop 10 Most Liked Reviews:")
print(top_reviews.to_string(index=False))


# ------------------------------------------------------------
# 13. AVERAGE THUMBS-UP BY RATING
# ------------------------------------------------------------

rating_thumbs = (
    df.groupby('score')['thumbsUpCount']
    .mean()
)

print("\nAverage Thumbs-Up by Rating:")
print(rating_thumbs)


plt.figure(figsize=(8, 5))

rating_thumbs.plot(kind='bar')

plt.title("Average Thumbs-Up Count by Rating")
plt.xlabel("Rating")
plt.ylabel("Average Thumbs-Up Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 14. TOP APP VERSIONS BY REVIEW COUNT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("8. APP VERSION ANALYSIS")
print("=" * 70)

version_counts = (
    df['reviewCreatedVersion']
    .value_counts()
    .head(10)
)

print("\nTop 10 App Versions:")
print(version_counts)


plt.figure(figsize=(10, 5))

version_counts.plot(kind='bar')

plt.title("Top 10 Google Pay App Versions by Review Count")
plt.xlabel("App Version")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 15. AVERAGE RATING BY APP VERSION
# ------------------------------------------------------------

version_rating = (
    df.groupby('reviewCreatedVersion')['score']
    .agg(['count', 'mean'])
    .sort_values('count', ascending=False)
    .head(10)
)

print("\nTop 10 App Versions - Review Count & Average Rating:")
print(version_rating)


# ------------------------------------------------------------
# 16. GOOGLE PAY RESPONSE ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("9. GOOGLE PAY RESPONSE ANALYSIS")
print("=" * 70)

df['has_reply'] = df['replyContent'].notna()

reply_counts = df['has_reply'].value_counts()

print("Reviews With/Without Reply:")
print(reply_counts)


# Response percentage
response_rate = df['has_reply'].mean() * 100

print(
    "\nGoogle Pay Response Rate:",
    round(response_rate, 2),
    "%"
)


# Response chart
reply_labels = ['No Reply', 'Replied']

reply_values = [
    (~df['has_reply']).sum(),
    df['has_reply'].sum()
]

plt.figure(figsize=(8, 5))

plt.bar(reply_labels, reply_values)

plt.title("Google Pay Review Response Status")
plt.xlabel("Response Status")
plt.ylabel("Number of Reviews")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 17. RESPONSE RATE BY RATING
# ------------------------------------------------------------

response_by_rating = (
    df.groupby('score')['has_reply']
    .mean()
    .mul(100)
    .round(2)
)

print("\nResponse Rate by Rating:")
print(response_by_rating)


plt.figure(figsize=(8, 5))

response_by_rating.plot(kind='bar')

plt.title("Google Pay Response Rate by Rating")
plt.xlabel("Rating")
plt.ylabel("Response Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 18. REVIEWS WITH MOST THUMBS-UP BY SENTIMENT
# ------------------------------------------------------------

sentiment_thumbs = (
    df.groupby('sentiment')['thumbsUpCount']
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage Thumbs-Up by Sentiment:")
print(sentiment_thumbs)


# ------------------------------------------------------------
# 19. POSITIVE / NEGATIVE REVIEW EXAMPLES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("10. REVIEW EXAMPLES")
print("=" * 70)

print("\nPositive Review Examples:")
positive_reviews = df[
    df['sentiment'] == 'Positive'
][['score', 'content']].head(5)

print(positive_reviews.to_string(index=False))


print("\nNegative Review Examples:")
negative_reviews = df[
    df['sentiment'] == 'Negative'
][['score', 'content']].head(5)

print(negative_reviews.to_string(index=False))


# ------------------------------------------------------------
# 20. MOST REVIEWED APP VERSIONS WITH RATING
# ------------------------------------------------------------

version_analysis = (
    df.groupby('reviewCreatedVersion')
    .agg(
        Review_Count=('score', 'count'),
        Average_Rating=('score', 'mean'),
        Average_Thumbs_Up=('thumbsUpCount', 'mean')
    )
    .sort_values(
        'Review_Count',
        ascending=False
    )
    .head(15)
)

version_analysis['Average_Rating'] = (
    version_analysis['Average_Rating']
    .round(2)
)

version_analysis['Average_Thumbs_Up'] = (
    version_analysis['Average_Thumbs_Up']
    .round(2)
)

print("\nApp Version Analysis:")
print(version_analysis)


# ------------------------------------------------------------
# 21. MONTHLY SENTIMENT ANALYSIS
# ------------------------------------------------------------

monthly_sentiment = pd.crosstab(
    df['year_month'],
    df['sentiment']
)

print("\nMonthly Sentiment:")
print(monthly_sentiment)


# ------------------------------------------------------------
# 22. TOP 10 MOST REVIEWED MONTHS
# ------------------------------------------------------------

top_months = (
    monthly_reviews
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Months by Review Count:")
print(top_months)


# ------------------------------------------------------------
# 23. FINAL PROJECT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("GOOGLE PAY INDIA - PROJECT SUMMARY")
print("=" * 70)

total_reviews = len(df)
average_rating = df['score'].mean()
five_star = (df['score'] == 5).sum()
one_star = (df['score'] == 1).sum()
positive = (df['sentiment'] == 'Positive').sum()
neutral = (df['sentiment'] == 'Neutral').sum()
negative = (df['sentiment'] == 'Negative').sum()
replied = df['has_reply'].sum()

print("Total Reviews          :", total_reviews)
print("Average Rating         :", round(average_rating, 2))
print("5-Star Reviews         :", five_star)
print("1-Star Reviews         :", one_star)
print("Positive Reviews       :", positive)
print("Neutral Reviews        :", neutral)
print("Negative Reviews       :", negative)
print("Reviews With Reply     :", replied)
print("Response Rate (%)      :", round(response_rate, 2))
print("Maximum Thumbs-Up      :", df['thumbsUpCount'].max())


# ------------------------------------------------------------
# 24. CREATE EDA SUMMARY TABLE
# ------------------------------------------------------------

summary = pd.DataFrame({
    'Metric': [
        'Total Reviews',
        'Average Rating',
        'Median Rating',
        'Most Common Rating',
        '5-Star Reviews',
        '1-Star Reviews',
        'Positive Reviews',
        'Neutral Reviews',
        'Negative Reviews',
        'Reviews With Reply',
        'Response Rate (%)',
        'Maximum Thumbs-Up'
    ],

    'Value': [
        total_reviews,
        round(average_rating, 2),
        median_rating,
        mode_rating,
        five_star,
        one_star,
        positive,
        neutral,
        negative,
        replied,
        round(response_rate, 2),
        df['thumbsUpCount'].max()
    ]
})

print("\n" + "=" * 70)
print("EDA SUMMARY TABLE")
print("=" * 70)

print(summary.to_string(index=False))


# ------------------------------------------------------------
# 25. SAVE EDA DATASET
# ------------------------------------------------------------

df.to_csv(
    "GooglePayIndia_EDA.csv",
    index=False
)

summary.to_csv(
    "GooglePayIndia_EDA_Summary.csv",
    index=False
)

print("\n" + "=" * 70)
print("STEP 2 COMPLETED SUCCESSFULLY")
print("=" * 70)

print("Files created:")
print("1. GooglePayIndia_EDA.csv")
print("2. GooglePayIndia_EDA_Summary.csv")
```
