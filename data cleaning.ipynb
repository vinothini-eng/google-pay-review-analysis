```python
# ============================================================
# GOOGLE PAY INDIA DATA ANALYSIS
# STEP 1: DATA CLEANING
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. Load Dataset
# ------------------------------------------------------------
from google.colab import files

uploaded = files.upload()
df = pd.read_csv('GooglePayIndia.csv')

print("=" * 60)
print("1. DATASET INFORMATION")
print("=" * 60)

print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

print("\nFirst 5 rows:")
print(df.head())


# ------------------------------------------------------------
# 2. Check Column Names
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("2. COLUMN NAMES")
print("=" * 60)

print(df.columns.tolist())


# ------------------------------------------------------------
# 3. Check Data Types
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("3. DATA TYPES")
print("=" * 60)

print(df.dtypes)


# ------------------------------------------------------------
# 4. Check Missing Values
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("4. MISSING VALUES")
print("=" * 60)

missing_values = df.isnull().sum()

print(missing_values)

print("\nMissing value percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))


# ------------------------------------------------------------
# 5. Check Duplicate Rows
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("5. DUPLICATE ROWS")
print("=" * 60)

duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)


# ------------------------------------------------------------
# 6. Check Duplicate Review IDs
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("6. DUPLICATE REVIEW IDs")
print("=" * 60)

duplicate_review_ids = df['reviewId'].duplicated().sum()

print("Duplicate review IDs:", duplicate_review_ids)


# ------------------------------------------------------------
# 7. Check Rating Values
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("7. RATING CHECK")
print("=" * 60)

print("Unique ratings:")
print(sorted(df['score'].dropna().unique()))

print("\nRating distribution:")
print(df['score'].value_counts().sort_index())

# Find invalid ratings
invalid_scores = df[
    ~df['score'].isin([1, 2, 3, 4, 5])
]

print("\nInvalid ratings:", len(invalid_scores))


# ------------------------------------------------------------
# 8. Check Review Text
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("8. REVIEW TEXT CHECK")
print("=" * 60)

missing_content = df['content'].isnull().sum()

empty_content = (
    df['content']
    .fillna('')
    .str.strip()
    .eq('')
    .sum()
)

print("Missing review text :", missing_content)
print("Empty review text   :", empty_content)


# ------------------------------------------------------------
# 9. Check Thumbs Up Count
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("9. THUMBS UP COUNT CHECK")
print("=" * 60)

print(df['thumbsUpCount'].describe())

negative_thumbs = df[
    df['thumbsUpCount'] < 0
]

print("\nNegative thumbs-up values:", len(negative_thumbs))


# ------------------------------------------------------------
# 10. Convert Date Columns
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("10. DATE CONVERSION")
print("=" * 60)

df['at'] = pd.to_datetime(
    df['at'],
    errors='coerce'
)

df['repliedAt'] = pd.to_datetime(
    df['repliedAt'],
    errors='coerce'
)

print("Date columns converted successfully.")


# ------------------------------------------------------------
# 11. Check Invalid Dates
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("11. INVALID DATE CHECK")
print("=" * 60)

invalid_review_dates = df['at'].isna().sum()
invalid_reply_dates = df['repliedAt'].isna().sum()

print("Missing/invalid review dates :", invalid_review_dates)
print("Missing/invalid reply dates  :", invalid_reply_dates)


# ------------------------------------------------------------
# 12. Check Reply Date Consistency
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("12. REPLY DATE CONSISTENCY")
print("=" * 60)

invalid_reply_date_rows = df[
    df['repliedAt'].notna() &
    df['at'].notna() &
    (df['repliedAt'] < df['at'])
]

print(
    "Reply date earlier than review date:",
    len(invalid_reply_date_rows)
)


# ------------------------------------------------------------
# 13. Remove Unnecessary Columns
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("13. REMOVE UNNECESSARY COLUMNS")
print("=" * 60)

columns_to_remove = [
    'Unnamed: 0',
    'userImage'
]

# Remove only columns that actually exist
columns_to_remove = [
    col for col in columns_to_remove
    if col in df.columns
]

df.drop(
    columns=columns_to_remove,
    inplace=True
)

print("Removed columns:", columns_to_remove)


# ------------------------------------------------------------
# 14. Remove Rows with Missing Review Text
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("14. REMOVE MISSING REVIEW TEXT")
print("=" * 60)

before_rows = len(df)

df.dropna(
    subset=['content'],
    inplace=True
)

after_rows = len(df)

print("Rows before:", before_rows)
print("Rows after :", after_rows)
print("Rows removed:", before_rows - after_rows)


# ------------------------------------------------------------
# 15. Remove Completely Duplicate Rows
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("15. REMOVE DUPLICATES")
print("=" * 60)

before_duplicates = len(df)

df.drop_duplicates(
    inplace=True
)

after_duplicates = len(df)

print("Rows before:", before_duplicates)
print("Rows after :", after_duplicates)
print("Duplicates removed:", before_duplicates - after_duplicates)


# ------------------------------------------------------------
# 16. Final Data Types
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("16. FINAL DATA TYPES")
print("=" * 60)

print(df.dtypes)


# ------------------------------------------------------------
# 17. Final Missing Values
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("17. FINAL MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())


# ------------------------------------------------------------
# 18. Final Dataset Information
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("18. FINAL DATASET")
print("=" * 60)

print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

print("\nFinal columns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 19. Display Cleaned Data
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("19. CLEANED DATA - FIRST 10 ROWS")
print("=" * 60)

print(df.head(10))


# ------------------------------------------------------------
# 20. Save Cleaned Dataset
# ------------------------------------------------------------

df.to_csv(
    "GooglePayIndia_Cleaned.csv",
    index=False
)

print("\n" + "=" * 60)
print("DATA CLEANING COMPLETED")
print("=" * 60)

print("Cleaned file saved as:")
print("GooglePayIndia_Cleaned.csv")
```
